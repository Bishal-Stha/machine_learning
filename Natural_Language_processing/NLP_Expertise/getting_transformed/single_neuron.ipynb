```python
import tensorflow as tf
from tensorflow.keras import layers
```

```python
x = tf.constant([[1.24, 3.85, 2.53, 5.91]])
w = tf.constant([[0.75],[0.75],[0.75],[0.75]])
b = tf.constant([[2.15]])
x.shape, w.shape, b.shape
```

```python
z = tf.matmul(x,w)+b
y = tf.nn.relu(z)
print(y)
```

```python
### Now lets do it using Keras and tensorflow.
model = layers.Dense(
    1,
    activation='relu',
    input_shape=(4,)
)
```

```python
x = tf.constant([[1.24, 3.85, 2.53, 5.91]])
_ = model(x)
```

```python
model.set_weights([
    w.numpy(),           # shape (4, 1)
    b.numpy().reshape(1) # shape (1,)
])
```
