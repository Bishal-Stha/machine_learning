```python
import tensorflow as tf
```

```python
a = tf.constant(7)
b = tf.constant([1,3,5])
c = tf.constant([[1,3,5],[2,4,6]])
print(f"{a}\n\n {b}\n\n{c}")
```

```python
x = tf.constant([1,2,3],dtype=tf.float32)
x
```

```python
zeros = tf.zeros((2,3))
ones = tf.ones((3,2))
random = tf.random.normal((2,2))
print(f"{zeros}\n\n {ones}\n\n{random}")
```

```python
print(f"{random.shape}, {random.dtype}, {tf.rank(random)}")
```

```python
p = tf.constant([[1,2],[3,4]])
q = tf.constant([[4,3],[2,1]])
print("Matrix Multiplication: ",tf.matmul(p,q))
print("Element wise mulitplication: ",(p*q))
```

```python
x = tf.constant([[1,2,3]],dtype=tf.float32) # (1*3)
w = tf.constant([0.5,0.5,0.5]) # (3,)

x*w
```

```python
## VVI
x = tf.constant([1,2,3,4])
print(tf.reshape(x,(2,2)))
print(tf.cast(x,tf.float32))
```

```python
import numpy as np
# Numpy to tensorflow
n = np.array([1,3,5])
t = tf.convert_to_tensor(n)
print(t)
t.numpy()
```
