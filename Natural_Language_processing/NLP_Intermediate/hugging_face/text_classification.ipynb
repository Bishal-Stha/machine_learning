```python
from transformers import pipeline

# Sentiment Analysis
pipe = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english",
)
```

```python
print(pipe("Wi-Fi is slower than snail today."))
```

```python
# Grammatical Correctness
grammar_checker = pipeline(
    task = "text-classification",
    model = "abdulmatinomotoso/English_Grammar_Checker"
)
```

```python
print(grammar_checker("He eat pizza every day."))
```

```python
# QNLI: Question Answering Natural Language Interface
qnli = pipeline(
    task = "text-classification",
    model = "cross-encoder/qnli-electra-base"   
)
```

```python
qnli("Where is Itahari located?, Itahari is located in Sunsari.")
```

```python
from transformers import pipeline
```

```python
# Dynamic category assignment

classifier = pipeline(
    task = "zero-shot-classification",
    model ="facebook/bart-large-mnli"
)

text = "Hey, Datacamp; we would like to feature your courses in our newsletter!"
categories = ["marketting","sales","support"]

ouptuts = classifier(text,categories)
```

```python
print(f"Top Label: {ouptuts['labels'][0]} with score: {ouptuts['scores'][0]}")
```
