```python
import spacy
```

```python
nlp = spacy.load("en_core_web_md")
```

```python
print("Some details about word vectors in spaCy:")
print(nlp.meta["lang"])
```

```python
for x in nlp.meta:
    if type(nlp.meta[x]) not in [list, dict]:
        print(f"{x}: {nlp.meta[x]}")
```

```python
nlp.meta["vectors"]
for x in nlp.meta["vectors"]:
    print(f"{x}: {nlp.meta['vectors'][x]}")
```

```python
print(f"NLP Vocab: {nlp.vocab}")
vocab_id = nlp.vocab["like"]
print(f"Vocab ID for 'like': {vocab_id}")
print(f"NLP Vocab Strings: {nlp.vocab.strings}")
vocab_string_id = nlp.vocab.strings["like"]
print(f"Vocab String ID for 'like': {vocab_string_id}")
print(f"NLP vocab vectors of like string: {nlp.vocab.vectors[vocab_string_id]}, length: {len(nlp.vocab.vectors[vocab_string_id])}")
```
