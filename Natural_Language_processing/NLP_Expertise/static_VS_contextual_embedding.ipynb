## Static Embeddings:
- Word2Vec
- FastText
- GloVe

```python
# Example with Gensim Word2Vec
from gensim.models import Word2Vec

sentences = [["I", "love", "NLP"], ["Word2Vec", "is", "cool"]]
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1)
vector = model.wv['NLP']  # get embedding of a word
similar = model.wv.most_similar('NLP')  # find similar words
```
