```python
import speech_recognition as sr

recognizer = sr.Recognizer()
recognizer
```

```python
with sr.AudioFile("./charlie-bit-me-5.wav") as source:
    charlie_bit_me = recognizer.record(source)
charlie_bit_me
```

```python
text = recognizer.recognize_google(charlie_bit_me,language="en-US")
text
```

```python
with sr.AudioFile("./good-morning-japanense.wav") as source:
    japanese = recognizer.record(source)
japanese
```

```python
text2 = recognizer.recognize_google(japanese, language="ja")
text2
```

```python
# Load or record the audio data
with sr.AudioFile("./leopard.wav") as source:
    leopard = recognizer.record(source)
```

```python
try:
    text3 = recognizer.recognize_google(leopard, 
                                        language="en-US", 
                                        show_all=True)
except sr.UnknownValueError:
    text3 = "Could not understand audio"
text3
```
