```python
def create_text_list(folder):
    text_list = []
    for file in folder:
        if file.endswith(".wav"):
            text = transcribe_audio(file)
            text_list.append(text)
    return text_list
```

```python
post_purchase_text = transcribe_audio("post-purchase-audio-27.wav")
pre_purchase_text =  transcribe_audio("pre-purchase-audio-25.wav")
```

```python
import pandas as pd

post_purchase_df = pd.DataFrame([{"label":"post_purchase","text":post_purchase_text}])
pre_purchase_df = pd.DataFrame([{"label":"pre_purchase","text":pre_purchase_text}])
df = pd.concat([post_purchase_df,pre_purchase_df], ignore_index=True)
df.head()
```

```python
# text classifier
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df["text"],
    df["label"],
    test_size=0.5, random_state=42
)

```

```python
# create text classifier pipeline

text_classifier = Pipeline([
    ("vectorizer",CountVectorizer()),
    ("tfidf",TfidfTransformer()),
    ("classifier",MultinomialNB())
])
```

```python
text_classifier.fit(X_train,y_train)
predict = text_classifier.predict(X_test)
accuracy = 100 * np.mean(predict == y_test)
print(f"The model {accuracy:.2f}% accurate.")
```
