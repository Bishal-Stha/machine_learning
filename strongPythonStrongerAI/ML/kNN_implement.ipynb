```python
class KNNClassifier:
    def __init__(self, k=3):
        self.k = k
        self.x_train = []
        self.y_train = []

    def fit(self, x_train, y_train):
        self.x_train = x_train
        self.y_train = y_train

    def predict(self, x_test):
        predictions = []
        for x in x_test:
            predictions.append(self._predict_one(x))
        return predictions

    def _predict_one(self, x):
        # 1. Calculate distances
        distances = []
        for i in range(len(self.x_train)):
            d = abs(self.x_train[i] - x)   # 1D absolute distance
            distances.append((d, self.y_train[i]))

        # 2. Sort by distance
        distances.sort(key=lambda tup: tup[0])

        # 3. Pick k nearest labels
        k_nearest = [label for (_, label) in distances[:self.k]]

        # 4. Count votes manually
        vote_count = {}
        for label in k_nearest:
            if label not in vote_count:
                vote_count[label] = 0
            vote_count[label] += 1

        # 5. Argmax: pick label with highest count
        best_label = None
        max_votes = -1
        for label, count in vote_count.items():
            if count > max_votes:
                max_votes = count
                best_label = label

        return best_label
```

```python
x_train = [1, 2, 3, 4, 5]
y_train = ['a', 'b', 'a', 'a', 'b']

model = KNNClassifier(k=3)
model.fit(x_train, y_train)

x_test = [3.5, 1.2, 4.8]
predictions = model.predict(x_test)

print("Predictions:", predictions)
# Output: ['a', 'a', 'b']
```
