```python
import pandas as pd

movie = pd.read_csv("../../NLP_Projects/IMDB Dataset.csv")
movie.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# min_n, max_n = 1,1

vect = CountVectorizer(max_features=100,
                        # ngram_range=(min_n,max_n)
                        )
vect.fit(movie.review)

```

```python
# transform the review column
X_review = vect.transform(movie.review)
type(X_review)
```

```python
X_review[:5]
```

```python
X_df = pd.DataFrame(X_review.toarray(),columns=vect.get_feature_names_out())
X_df.head()
```

```python
vect2 = CountVectorizer(max_df=50)
vect2.fit(movie.review)

X_review2 = vect2.transform(movie.review)
```

```python
# Only convert the first 5 rows to avoid MemoryError
X_df2 = pd.DataFrame(X_review2[:5].toarray(), columns=vect2.get_feature_names_out())

X_df2.head()
```
