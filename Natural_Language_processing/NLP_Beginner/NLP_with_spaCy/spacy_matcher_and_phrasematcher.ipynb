```python
import spacy
from spacy.matcher import Matcher

nlp = spacy.load("en_core_web_sm")
matcher = Matcher(nlp.vocab)

# Define pattern
pattern = [{"LOWER": "good"}, {"LOWER": "morning"}]
matcher.add("GOOD_MORNING", [pattern])

doc = nlp("He said good morning to everyone.")

matches = matcher(doc)
for match_id, start, end in matches:
    span = doc[start:end]
    print(span.text)  # Output: good morning
```

```python
import spacy
from spacy.matcher import PhraseMatcher

nlp = spacy.load("en_core_web_sm")
phrase_matcher = PhraseMatcher(nlp.vocab) # type: ignore

# List of phrases to match
phrases = ["New York", "San Francisco", "Los Angeles"]
patterns = [nlp(text) for text in phrases]

phrase_matcher.add("CITIES", patterns)

doc = nlp("I have visited New York and Los Angeles.")

matches = phrase_matcher(doc)
for match_id, start, end in matches:
    span = doc[start:end]
    print(span.text)  # Output: New York, Los Angeles
```
