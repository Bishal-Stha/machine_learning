```python
## Single Neuron Forward pass in tensorflow.
import tensorflow as tf
```

```python
x = tf.constant([[1.0,5.0,3.0]]) # shape: (1,3)
w = tf.constant([[0.5],[0.5],[0.5]]) # shape: (3,1)
b = tf.constant([[1.25]]) # shape: (1,1)
```

```python
z = tf.matmul(x,w)+b
print(z)
```

```python
# ReLU activation function
y = tf.nn.relu(z)
y
```

#### Same task as above but using Keras.

```python
from tensorflow.keras import layers
```

```python
neuron = layers.Dense(
    units=1,
    activation='relu',
    input_shape=(3,) # 3 input features
)
```

```python
# Input tensor
x = tf.constant([[1.0, 5.0,3.0]]) # shape: (1,3)
```

```python
# forward pass
y = neuron(x)
print(y.numpy())
```

```python
# neuron.set_weights([
#     [[0.5],[0.5],[0.5]], # w (3,1)
#     [1.25] # b (1,)
# ])
neuron.set_weights([
    tf.constant([[0.5], [0.5], [0.5]]),  # W (3,1)
    tf.constant([1.25])                 # b (1,)
])

y = neuron(x)
print(y.numpy())
```
