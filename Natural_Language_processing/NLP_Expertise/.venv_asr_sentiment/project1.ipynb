```python
import os
from pydub import AudioSegment
import tempfile
import pyttsx3
from transformers import pipeline
```

```python
# 1️⃣ Initialize pipelines
asr_pipeline = pipeline("automatic-speech-recognition", model="openai/whisper-tiny")
sentiment_pipeline = pipeline("sentiment-analysis")

# 2️⃣ Initialize TTS engine
tts_engine = pyttsx3.init()
```

```python
def convert_to_wav(audio_path):
    """Convert any audio format to wav for consistency."""
    ext = os.path.splitext(audio_path)[1].lower()
    if ext == ".wav":
        return audio_path
    audio = AudioSegment.from_file(audio_path)
    tmp_wav = tempfile.NamedTemporaryFile(delete=False, suffix=".wav").name
    audio.export(tmp_wav, format="wav")
    return tmp_wav
```

```python
def transcribe_audio(audio_path):
    wav_path = convert_to_wav(audio_path)
    result = asr_pipeline(wav_path)
    return result["text"]
```

```python

def analyze_sentiment(text):
    result = sentiment_pipeline(text)[0]
    return result["label"], result["score"]
```

```python
def speak_text(text):
    tts_engine.say(text)
    tts_engine.runAndWait()
```

```python
def process_audio(audio_path):
    print(f"Processing audio: {audio_path}")
    text = transcribe_audio(audio_path)
    print(f"Transcribed Text: {text}")
    label, score = analyze_sentiment(text)
    sentiment_text = f"Sentiment: {label} (confidence: {(score*100):.2f}) %"
    print(sentiment_text)
    speak_text(sentiment_text)
```

```python
if __name__ == "__main__":
    # Replace with your audio file path
    audio_file = str(input("enter audio file name: "))
    process_audio(audio_file)
```
