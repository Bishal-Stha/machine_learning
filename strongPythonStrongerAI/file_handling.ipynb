Something cool happens here because **r+ points to the beginning of the file and expects to be read first and written later. but when we write first and read later. It can create inconsistency**.

```python
with open("batman.jpg", "rb") as pic:
    data = pic.read()

with open("batman_copy.jpg", "wb") as pic2:
    pic2.write(data)

pic.close()
pic2.close()
```

```python
def read_file():
    try:
        filePath: str = input("Enter filename: ")
        with open(f"{filePath}.txt", "r") as fname:  # using with ensures auto-closing
            print(fname.read())
    except FileNotFoundError:
        print("Error: The file does not exist.")
    except PermissionError:
        print("Error: You do not have permission to read this file.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```

```python
read_file()
```
