```python
setA = {'a','b','c','d','e','e','e'}
setB = {'a','e','i','o','u'}

AUB = setA.union(setB)
print(sorted(AUB))
```

```python
AnB = setA.intersection(setB)
print(sorted(AnB))
```

```python
AdifB = setA.difference(setB)
print(sorted(AdifB))

BdifA = setB.difference(setA)
print(sorted(BdifA))
```
