## Stuffs with Wave Library

```python
import wave

audio = wave.open("good_morning.wav","r")
audio
```

```python
audio_binary = audio.readframes(-1)
audio_binary[:10]
```

```python
import numpy as np
```

```python
audio_int = np.frombuffer(audio_binary,dtype=np.int16)
audio_int[:10]
```

```python
audio_frame_rate = audio.getframerate()
audio_frame_rate
```

```python
## Stuffs with speech_recognition
import speech_recognition as sr

recognizer = sr.Recognizer()

with sr.AudioFile("clean-support-call.wav") as source:
    audio_data = recognizer.record(source)

type(audio_data)
```

```python
# text from the audio.
text = recognizer.recognize_google(audio_data,language="en-US")
text
```

## some limitations of speech recognition. As it is still in its early phase and many innovation is yet to happen in the domain.

```python
with sr.AudioFile("leopard.wav") as source:
    leopard_audio = recognizer.record(source)

text2 = recognizer.recognize_google(leopard_audio,language="en-US")
# This gives UnknownValueError as leopard roar is a non-human language. 
```

## Stuffs with PyDub

```python
from pydub import AudioSegment

gm_wav = AudioSegment.from_file("good_morning.wav")
type(gm_wav)
```

```python
print("Channels: ",gm_wav.channels)
print("Duration: ",gm_wav.duration_seconds)
print("Max Amplitude: ",gm_wav.max)
print("Frame Rate: ",gm_wav.frame_rate)
print("Frame Width: ",gm_wav.frame_width)
print("Sample Width: ",gm_wav.sample_width)
```

```python
from pydub.playback import play
from pydub.effects import normalize

voice = AudioSegment.from_file("good-afternoon.wav")
type(voice)
play(voice)
```

```python
normalized_voice = normalize(voice)
play(normalized_voice)
```
