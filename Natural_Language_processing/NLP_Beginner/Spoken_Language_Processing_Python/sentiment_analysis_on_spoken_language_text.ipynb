```python
# Custom named entities
from spacy.pipeline import EntityRuler
```

```python
NLP = spacy.load("en_core_web_sm")
```

```python
# check spacy pipeline
print(NLP.pipeline)
```

```python
# Get the EntityRuler pipe and add patterns
ruler = NLP.get_pipe("entity_ruler")
ruler.add_patterns([
    {"label": "PRODUCT", "pattern": [{"LEMMA": "smartphone"}]},
    {"label": "PRODUCT", "pattern": [{"LOWER": "smart"}, {"LOWER": "watches"}]},
])
```

```python
NLP.pipeline
```

```python
doc = "Our main product are smartphones and smart watches. Our company was established in 1997 AD. Our company name is Google and I am Victor Von Doom."
```

```python
for entity in NLP(doc).ents:
    print(entity.text, entity.label_)
```
