```python
import spacy

nlp = spacy.load('en_core_web_sm')
print(nlp.get_pipe("ner"))

text = """On July 4, 2025, Amazon acquired a startup for $2.5 billion in Berlin. 
Barack Obama, the former president of the United States, gave a speech in Spanish 
at the Eiffel Tower. The event, called TechFuture Expo 2025, was attended by 
Norwegians and Democrats. They discussed Einstein's theory of relativity, 
the Mona Lisa, and Section 230 of the Communications Decency Act. 
By 8:00 PM, 75% of attendees had left the venue located at 221B Baker Street. 
The remaining guests spoke in Mandarin and ate Toblerone bars. 
Microsoft and NASA also participated, showcasing a new rocket that weighed 15 tons."""


doc = nlp(text) 
print(doc.ents)


```

```python
doc.ents[0].label_
```

```python
for i in range(len(doc.ents)):
    print(f"{doc.ents[i]}: {doc.ents[i].label_} / {doc.ents[i].label}")
```

```python
import stanza

# Download Spanish models (only once)
stanza.download('es')

# Initialize the Spanish pipeline
nlp = stanza.Pipeline('es')

# Sample Spanish text
doc = nlp("Hola, me llamo Bishal y estoy aprendiendo procesamiento de lenguaje natural.")

# Print sentences and their tokens
for i, sentence in enumerate(doc.sentences): # type: ignore
    print(f"Sentence {i+1}:")
    for word in sentence.words:
        print(f"  {word.text}\t{word.lemma}\t{word.upos}")

# Output example: tokens, lemmas, and POS tags
```
