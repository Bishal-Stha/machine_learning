```python
import nltk

# nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter

```

```python
text = "The cat is in the box. The cat likes the box. The box is over the cat."

tokens = [w for w in word_tokenize(text.lower()) if w.isalpha()]

no_stops = [t for t in tokens if t not in stopwords.words('english')]
Counter(tokens), Counter(no_stops), Counter(no_stops).most_common(1)
```

```python
from gensim.corpora.dictionary import Dictionary
from nltk.tokenize import word_tokenize
```

```python
my_documents = ['The movie was about spaceship and aliens',
                'I really liked the movie',
                'Awesome action scenes, but boring characters.',
                'The movie was awful, i hate alien movies,',
                'More space movies, please']

tokenized_words = [word_tokenize(doc.lower()) for doc in my_documents]

dicionary = Dictionary(tokenized_words)
dicionary.token2id
```

```python
corpus = [dicionary.doc2bow(doc) for doc in tokenized_words]
corpus
```
