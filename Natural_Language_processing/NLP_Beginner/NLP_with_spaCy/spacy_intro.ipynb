```python
import spacy
nlp = spacy.load("en_core_web_sm")
nlp
```

```python
text = "A spaCy pipeline object is created."
doc = nlp(text)
doc.text
```

```python
# Tokenization with spacy
print([token.text for token in doc])
```

```python
# sentence segmenation with spacy.
doc = nlp("My name is bishal shrestha. I am 18 years old.")
for sent in doc.sents:
    print(sent.text)
```

```python
# Lemmatization with spacy.
doc = nlp("When i woke up at 6:30, i got scolded by my mom.")
print([(token.text, token.lemma_) for token in doc])
```

Linguistic Features in spaCy

```python
# POS tagging: Part-Of-Speech tagging
text = "Albert Einstein was genius. He met Robert Openheimer in Manhattan in 1943 AD."
doc = nlp(text)

print([(token.text, token.ent_type_) for token in doc])

# print([])
```

```python
verb_sent = "I watch Television."

print([(token.text, token.pos_, spacy.explain(token.pos_)) for token in nlp(verb_sent)]) # type: ignore
```

```python
noun_sent = "I left without my watch."

print([(token.text, token.pos_, spacy.explain(token.pos_)) for token in nlp(noun_sent)]) # type: ignore
```
