```python
import spacy

nlp = spacy.load("en_core_web_md")
```

```python
text1 = "I live in Nepal."
text2 = "I am a Nepalese citizen."
doc1, doc2 = nlp(text1), nlp(text2)

print(f"Similarity between doc1 and doc2: {round(doc1.similarity(doc2),3)}")
```

```python
# span similarity
span1 = doc1[1:4]  # "live in Nepal"
span2 = doc2[3:6]  # "Neapali citizen"

print(f"Similarity between span1 and span2: {round(span1.similarity(span2),3)}")
```

```python
# doc similarity
doc1 = nlp("A quick brown fox jumps over the lazy dog.")
doc2 = nlp("A quick brown fox jumps over the lazy dog.")
print(f"Similarity between doc1 and doc2: {round(doc1.similarity(doc2),3)}")
```

```python
# sentence similarity
sentences = nlp("What is the cheapest flight from Boston to Seattle? Which airline serves Denver, Pittsburgh and Atlanta? What kinds of planes are used by American Airlines?")

keyword = nlp("price")

for i, sentence in enumerate(sentences.sents):
    print(f"Similarity between sentence {i+1} and keyword: {round(sentence.similarity(keyword),5)}")
```
