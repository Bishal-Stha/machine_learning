```python
# Default Pipeline for NER

import spacy

# Load the full default pipeline shipped with the model
nlp = spacy.load("en_core_web_sm")        # Tokenizer is built‑in

print("Pipeline order:", nlp.pipe_names)

text = "Apple is looking at buying a U.K. startup for $1 billion."

doc = nlp(text)

# Show what the default pipeline produced
print("\nTokens / POS / Lemma")
for tok in doc:
    print(f"{tok.text:<12} {tok.pos_:<6} {tok.lemma_}")

print("\nNamed Entities")
for ent in doc.ents:
    print(ent.text, ent.label_)

print("\nDependency Head → Child")
for tok in doc:
    print(tok.head.text, "→", tok.text)
```

```python
# Custom Pipeline for NER
import spacy

# Create a blank English NLP object
nlp = spacy.blank("en")

# Add a sentencizer for sentence boundary detection
nlp.add_pipe("sentencizer")

# Add a rule-based entity ruler
ruler = nlp.add_pipe("entity_ruler")
ruler.add_patterns([
    {"label": "TOOL", "pattern": "spaCy"},
    {"label": "HOBBY", "pattern": [{"LOWER": "coding"}]}
])

# Add the ML-based NER from pre-trained model (optional)
ner_model = spacy.load("en_core_web_sm")
nlp.add_pipe("ner", source=ner_model, last=True)

# Print the pipeline
print("Custom Pipeline Components:", nlp.pipe_names)

# Example text
text = "I enjoy coding and using spaCy. Google is also cool."

# Process the text
doc = nlp(text)

# Show detected sentences
print("\nSentences:")
for sent in doc.sents:
    print("-", sent.text)

# Show detected entities
print("\nNamed Entities:")
for ent in doc.ents:
    print(f"{ent.text:<20} {ent.label_}")
```
