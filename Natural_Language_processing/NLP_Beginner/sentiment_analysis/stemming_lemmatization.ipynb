```python
# Import libraries
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.stem.snowball import SnowballStemmer
```

```python
# Stemming
stemmer = PorterStemmer()
stemmer.stem("Learning")
```

```python
text = "today is a wonderful day."
stemmer.stem(text)
```

```python
text_tokens = word_tokenize(text)
stemed_text = [stemmer.stem(word) for word in text_tokens]
stemed_text
```

```python
# SnowballStemmer (Non-English Stemmer)
snowStemmer = SnowballStemmer("dutch")
snowStemmer.stem("Beginnen")
```

```python
# Lemmatizer
wordnet = WordNetLemmatizer()
wordnet.lemmatize("wonderful",pos='a')
```
