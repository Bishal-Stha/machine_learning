### **Complete flow of information.**
- Convert wav or any audio file format into wav_object.
- Now read frames of audio object.
- Then convert the binary type of sound object into integer type.
- Now, get the frame rate
- Then, get the timestamp. 

```python
import wave

# Importing audio file as wave object
good_morning = wave.open("good_morning.wav","r")

good_morning_wave = good_morning.readframes(-1)

good_morning_wave[:10]
```

```python
type(good_morning_wave)
```

```python
#converting byte to integers in numpy.
import numpy as np

# converting byte to integers in numpy
signal_gm = np.frombuffer(good_morning_wave, dtype=np.int16)

# Show the first 10 values
print("First 10 values of the signal:", signal_gm[:10])
# Show the type of the signal
print("Type of the signal:", type(signal_gm))
# Show the shape of the signal
print("Shape of the signal:", signal_gm.shape)
```

```python
# Get the frame rate
frame_rate_gm = good_morning.getframerate()
frame_rate_gm
```

```python
len(signal_gm) / frame_rate_gm
```

```python
# Getting the time stamp of the good morning sound wave
time_gm = np.linspace(start=0, stop= len(signal_gm)/frame_rate_gm, num=len(signal_gm))
time_gm[:10]
```

```python
# doing similar preprocessing with good afternoon also.
good_afternoon = wave.open("good-afternoon.wav","r")
good_afternoon_wav = good_afternoon.readframes(-1)
signal_ga = np.frombuffer(good_afternoon_wav,dtype='int16')

frame_rate_ga = good_afternoon.getframerate()
time_ga = np.linspace(start=0, stop= len(signal_ga)/frame_rate_ga, num=len(signal_ga))
```

```python
# Visualizing sound wave.
import matplotlib.pyplot as plt
plt.title("Good Morning v/s Good Afternoon")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.plot(time_gm, signal_gm, label="Good Morning")
plt.plot(time_ga,signal_ga, label="Good Afternoon", alpha=0.5)
plt.legend()
plt.show()
```
