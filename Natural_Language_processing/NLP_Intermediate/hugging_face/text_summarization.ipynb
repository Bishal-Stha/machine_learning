```python
from transformers import pipeline
```

```python
summarizer = pipeline(
    task = "summarization",
    model = "nyamuda/extractive-summarization",
    
)

text = "this is my really large text about Data Science..."
summary_text = summarizer(text)
```

```python
summary_text[0]["summary_text"]
```

```python
abs_summarizer = pipeline(
    task = "summarization",
    model= "sshleifer/distilbart-cnn-12-6"
)

abs_summary_text = abs_summarizer(text)
print(abs_summary_text[0]["summary_text"])
```
