### Transformer model provides following capabilities.
| Task                    | Model                                      |
|-------------------------|--------------------------------------------|
| Text Generation         | model: distilgpt2                          |
| Named Entity Recognition (NER) | model: dbmdz/bert-large-cased-finetuned-conll03-english |
| Sentiment Analysis      | model: distilbert-base-uncased-finetuned-sst-2-english |
| Question Answering      | model: deepset/roberta-base-squad2        |
| Fill-mask               | model: bert-base-uncased                   |
| Language Translation (EN→FR) | model: Helsinki-NLP/opus-mt-en-fr     |
| Zero-shot Classification| model: facebook/bart-large-mnli           |
| Summarization           | model: facebook/bart-large-cnn            |

```python
from transformers import pipeline

unmasker = pipeline("fill-mask")
unmasker("This course will teach you all about <mask> models.", top_k=5)
```

```python
ner = pipeline("ner",grouped_entities=True)
ner("My name is Sylvain and I work at Hugging Face in Brooklyn.")
```

```python
question_answerer = pipeline("question-answering",model="deepset/roberta-base-squad2")
question_answerer({
    "question": "Who am i ?",
    "context": "my name is Bishal Shrestha, i am 19 years old"
})
```

```python
from transformers import pipeline

image_classifier = pipeline(
    task="image-classification", model="google/vit-base-patch16-224", device=0
)
result = image_classifier(
    "batman.jpg"
)
print(result)
```

```python
from transformers import pipeline

transcriber = pipeline(
    task="automatic-speech-recognition", model="openai/whisper-large-v3", device =0
)
result = transcriber(
    "https://huggingface.co/datasets/Narsil/asr_dummy/resolve/main/mlk.flac"
)
print(result)
```
