```python
import spacy
from nltk.corpus import stopwords
```

```python
with open("texts.txt", "r", encoding="utf-8") as file:
    data = file.read()
    print(data[:300])
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp(data)

tokens = [token.text.lower() for token in doc if token.text.isalpha()]
tokens[:10]
```

```python
no_stop_tokens = [token for token in tokens if token not in stopwords.words('english')]
no_stop_tokens[:10]
```

```python
paragraph = ' '.join(no_stop_tokens)
```

```python
paragraph[:500]
```

```python
doc = nlp(paragraph)
print([(token.lemma_, token.ent_type_)for token in doc if token.ent_type_ != ""])
```
